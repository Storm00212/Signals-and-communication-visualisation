```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["figure.figsize"] = (12, 5)
plt.rcParams["axes.grid"] = True

def show_plot(title, xlabel="Time (s)", ylabel="Amplitude"):
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.grid(True)
    plt.show()
```

```python
# ============================================================
# PROBLEM 5.1-1
# Trapezoidal message
# ============================================================

wc = 1e8          # carrier angular frequency
kf = 1e5
kp = 25

# From the worked solution:
# 10^-3 = 3a
a = 1e-3 / 3

# We use one complete period = 4a
T = 4 * a

# Time axis
N = 10000
t = np.linspace(0, T, N)

# ------------------------------------------------------------
# Construct trapezoidal message
# ------------------------------------------------------------

m = np.zeros_like(t)

for i, ti in enumerate(t):

    tau = ti % T

    if tau < a:
        # Low level
        m[i] = -1

    elif tau < 2*a:
        # Rising from -1 to +1
        m[i] = -1 + 2*(tau-a)/a

    elif tau < 3*a:
        # High level
        m[i] = 1

    else:
        # Falling from +1 to -1
        m[i] = 1 - 2*(tau-3*a)/a


# ------------------------------------------------------------
# Derivative dm/dt
# ------------------------------------------------------------

dm_dt = np.zeros_like(t)

for i, ti in enumerate(t):

    tau = ti % T

    if a < tau < 2*a:
        dm_dt[i] = 2/a

    elif 3*a < tau < 4*a:
        dm_dt[i] = -2/a

    else:
        dm_dt[i] = 0


# ------------------------------------------------------------
# PM phase
# ------------------------------------------------------------

theta_PM = wc*t + kp*m


# ------------------------------------------------------------
# FM phase
# Numerical integration
# ------------------------------------------------------------

dt = t[1] - t[0]

integral_m = np.cumsum(m) * dt

theta_FM = wc*t + kf*integral_m


# ------------------------------------------------------------
# Instantaneous angular frequencies
# ------------------------------------------------------------

omega_PM = wc + kp*dm_dt
omega_FM = wc + kf*m


# ============================================================
# PLOTS
# ============================================================

plt.figure()
plt.plot(t*1000, m)
plt.title("5.1-1: Trapezoidal Message m(t)")
plt.xlabel("Time (ms)")
plt.ylabel("m(t)")
plt.grid()
plt.show()


plt.figure()
plt.plot(t*1000, theta_PM)
plt.title("5.1-1: PM Phase θ_PM(t)")
plt.xlabel("Time (ms)")
plt.ylabel("Phase (rad)")
plt.grid()
plt.show()


plt.figure()
plt.plot(t*1000, theta_FM)
plt.title("5.1-1: FM Phase θ_FM(t)")
plt.xlabel("Time (ms)")
plt.ylabel("Phase (rad)")
plt.grid()
plt.show()


plt.figure()
plt.plot(t*1000, omega_PM/1e6, label="PM")
plt.plot(t*1000, omega_FM/1e6, label="FM")
plt.title("5.1-1: Instantaneous Angular Frequency")
plt.xlabel("Time (ms)")
plt.ylabel("ωᵢ × 10⁶ rad/s")
plt.legend()
plt.grid()
plt.show()
```

```python
# ============================================================
# PROBLEM 5.1-2
# Sawtooth message
# ============================================================

fc = 1e6
wc = 2*np.pi*fc

kf = 2000*np.pi
kp = np.pi/2

T = 1e-3

# Plot several periods
t = np.linspace(0, 3*T, 20000)

# ------------------------------------------------------------
# Sawtooth message from -1 to +1
# ------------------------------------------------------------

tau = t % T

m = -1 + 2*tau/T


# ------------------------------------------------------------
# Derivative
# ------------------------------------------------------------

dm_dt = np.ones_like(t) * (2/T)


# ------------------------------------------------------------
# PM phase
# ------------------------------------------------------------

theta_PM = wc*t + kp*m


# ------------------------------------------------------------
# FM phase
# ------------------------------------------------------------

dt = t[1] - t[0]
integral_m = np.cumsum(m) * dt

theta_FM = wc*t + kf*integral_m


# ------------------------------------------------------------
# Instantaneous frequencies
# ------------------------------------------------------------

omega_PM = wc + kp*dm_dt
omega_FM = wc + kf*m


# ============================================================
# PLOTS
# ============================================================

plt.figure()
plt.plot(t*1000, m)
plt.title("5.1-2: Sawtooth Message")
plt.xlabel("Time (ms)")
plt.ylabel("m(t)")
plt.grid()
plt.show()


plt.figure()
plt.plot(t*1000, theta_PM)
plt.title("5.1-2: PM Phase")
plt.xlabel("Time (ms)")
plt.ylabel("Phase (rad)")
plt.grid()
plt.show()


plt.figure()
plt.plot(t*1000, theta_FM)
plt.title("5.1-2: FM Phase")
plt.xlabel("Time (ms)")
plt.ylabel("Phase (rad)")
plt.grid()
plt.show()


plt.figure()
plt.plot(t*1000, omega_PM/(2*np.pi), label="PM")
plt.plot(t*1000, omega_FM/(2*np.pi), label="FM")
plt.axhline(fc, linestyle="--", label="Carrier")
plt.title("5.1-2: Instantaneous Frequency")
plt.xlabel("Time (ms)")
plt.ylabel("Frequency (Hz)")
plt.legend()
plt.grid()
plt.show()
```

```python
# ============================================================
# PROBLEM 5.1-3
# ============================================================

A = 10
wc = 10000*np.pi

# Given signal
omega_signal = 13000*np.pi

t = np.linspace(-1, 1, 5000)

# ------------------------------------------------------------
# Actual angle-modulated signal
# ------------------------------------------------------------

s = A*np.cos(omega_signal*t)


# ------------------------------------------------------------
# PM interpretation
#
# m(t) = 3*pi*t
# ------------------------------------------------------------

m_PM = 3*np.pi*t

theta_PM = wc*t + m_PM

s_PM = A*np.cos(theta_PM)


# ------------------------------------------------------------
# FM interpretation
#
# m(t) = 3*pi
# ------------------------------------------------------------

m_FM = 3*np.pi*np.ones_like(t)

kf = 1000

theta_FM = wc*t + kf*m_FM

# Note:
# To reproduce the exact signal, kf*m_FM must equal
# 3000*pi*t, which isn't possible with constant m_FM.
#
# Therefore the important purpose here is to visualize
# the derived PM message and the given signal.
# ------------------------------------------------------------


plt.figure()
plt.plot(t, s)
plt.title("5.1-3: Given Angle-Modulated Signal")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.xlim(-0.005, 0.005)
plt.grid()
plt.show()


plt.figure()
plt.plot(t, m_PM)
plt.title("5.1-3: Message Obtained Assuming PM")
plt.xlabel("Time (s)")
plt.ylabel("m(t)")
plt.grid()
plt.show()


plt.figure()
plt.plot(t, m_FM)
plt.title("5.1-3: Message Obtained Assuming FM")
plt.xlabel("Time (s)")
plt.ylabel("m(t)")
plt.grid()
plt.show()
```

```python
# ============================================================
# PROBLEM 5.2-1
# Two-tone message
# ============================================================

A = 10
wc = 1e6

kf = 1000*np.pi
kp = 1

# Use a reasonable time scale for the message
t = np.linspace(0, 0.01, 20000)

# ------------------------------------------------------------
# Message
# ------------------------------------------------------------

m = 2*np.cos(100*t) + 18*np.cos(2000*np.pi*t)


# ------------------------------------------------------------
# PM
# ------------------------------------------------------------

theta_PM = wc*t + kp*m

s_PM = A*np.cos(theta_PM)


# ------------------------------------------------------------
# FM
# ------------------------------------------------------------

# Integral of:
# 2 cos(100t) = 0.02 sin(100t)
#
# 18 cos(2000πt)
# = 18/(2000π) sin(2000πt)

integral_m = (
    0.02*np.sin(100*t)
    +
    (18/(2000*np.pi))*np.sin(2000*np.pi*t)
)

theta_FM = wc*t + kf*integral_m

s_FM = A*np.cos(theta_FM)


# ------------------------------------------------------------
# Plot message
# ------------------------------------------------------------

plt.figure()
plt.plot(t, m)
plt.title("5.2-1: Two-Tone Message m(t)")
plt.xlabel("Time (s)")
plt.ylabel("m(t)")
plt.grid()
plt.show()


# ------------------------------------------------------------
# Plot PM
# ------------------------------------------------------------

plt.figure()
plt.plot(t, s_PM)
plt.title("5.2-1: PM Signal")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.xlim(0, 0.002)
plt.grid()
plt.show()


# ------------------------------------------------------------
# Plot FM
# ------------------------------------------------------------

plt.figure()
plt.plot(t, s_FM)
plt.title("5.2-1: FM Signal")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.xlim(0, 0.002)
plt.grid()
plt.show()
```

```python
# ============================================================
# PROBLEM 5.2-2
# ============================================================

A = 10
fc = 1e6
wc = 2*np.pi*fc

fm = 1000
wm = 2*np.pi*fm

beta = 0.1

# ------------------------------------------------------------
# Use a short time interval
# so that individual carrier cycles are visible
# ------------------------------------------------------------

t = np.linspace(0, 0.002, 100000)

# Modulated signal
s = A*np.cos(
    wc*t + beta*np.sin(wm*t)
)

# Message
m = np.sin(wm*t)

# Instantaneous frequency
omega_i = wc + beta*wm*np.cos(wm*t)

f_i = omega_i/(2*np.pi)


# ------------------------------------------------------------
# Message
# ------------------------------------------------------------

plt.figure()
plt.plot(t*1000, m)
plt.title("5.2-2: Message Signal")
plt.xlabel("Time (ms)")
plt.ylabel("m(t)")
plt.grid()
plt.show()


# ------------------------------------------------------------
# Instantaneous frequency
# ------------------------------------------------------------

plt.figure()
plt.plot(t*1000, f_i)
plt.axhline(fc, linestyle="--", label="Carrier frequency")

plt.title("5.2-2: Instantaneous Frequency")
plt.xlabel("Time (ms)")
plt.ylabel("Frequency (Hz)")
plt.legend()
plt.grid()
plt.show()


# ------------------------------------------------------------
# Zoom into carrier
# ------------------------------------------------------------

plt.figure()
plt.plot(t[:3000]*1e6, s[:3000])

plt.title("5.2-2: Zoomed Angle-Modulated Carrier")
plt.xlabel("Time (μs)")
plt.ylabel("Amplitude")
plt.grid()
plt.show()
```

```python
# ============================================================
# PROBLEM 5.2-3
# ============================================================

A = 5

fc = 1e6
wc = 2*np.pi*fc

w1 = 1000*np.pi
w2 = 2000*np.pi

t = np.linspace(0, 0.004, 100000)

# ------------------------------------------------------------
# Phase
# ------------------------------------------------------------

phase_deviation = (
    20*np.sin(w1*t)
    +
    10*np.sin(w2*t)
)

theta = wc*t + phase_deviation

s = A*np.cos(theta)


# ------------------------------------------------------------
# Message-equivalent phase deviation
# ------------------------------------------------------------

plt.figure()
plt.plot(t*1000, phase_deviation)

plt.title("5.2-3: Phase Deviation")
plt.xlabel("Time (ms)")
plt.ylabel("Phase deviation (rad)")
plt.grid()
plt.show()


# ------------------------------------------------------------
# Instantaneous frequency
# ------------------------------------------------------------

omega_i = (
    wc
    +
    20*w1*np.cos(w1*t)
    +
    10*w2*np.cos(w2*t)
)

f_i = omega_i/(2*np.pi)

plt.figure()
plt.plot(t*1000, f_i)

plt.axhline(fc, linestyle="--", label="Carrier")

plt.title("5.2-3: Instantaneous Frequency")
plt.xlabel("Time (ms)")
plt.ylabel("Frequency (Hz)")
plt.legend()
plt.grid()
plt.show()


# ------------------------------------------------------------
# Zoomed carrier
# ------------------------------------------------------------

plt.figure()
plt.plot(t[:3000]*1e6, s[:3000])

plt.title("5.2-3: Zoomed Modulated Carrier")
plt.xlabel("Time (μs)")
plt.ylabel("Amplitude")
plt.grid()
plt.show()
```

```python
# ============================================================
# PROBLEM 5.2-6
# ============================================================

kf = 20000*np.pi
kp = 10

fm = 1000
wm = 2*np.pi*fm

t = np.linspace(0, 0.005, 50000)


# ============================================================
# CASE 1: Normal
# ============================================================

m1 = np.sin(wm*t)

dm1_dt = wm*np.cos(wm*t)

f_FM_1 = (kf*m1)/(2*np.pi)
f_PM_1 = (kp*dm1_dt)/(2*np.pi)


# ============================================================
# CASE 2: Double amplitude
# ============================================================

m2 = 2*np.sin(wm*t)

dm2_dt = 2*wm*np.cos(wm*t)

f_FM_2 = (kf*m2)/(2*np.pi)
f_PM_2 = (kp*dm2_dt)/(2*np.pi)


# ============================================================
# CASE 3: Double frequency
# ============================================================

wm3 = 2*wm

m3 = np.sin(wm3*t)

dm3_dt = wm3*np.cos(wm3*t)

f_FM_3 = (kf*m3)/(2*np.pi)
f_PM_3 = (kp*dm3_dt)/(2*np.pi)


# ============================================================
# PLOTS
# ============================================================

plt.figure()
plt.plot(t*1000, m1, label="Normal")
plt.plot(t*1000, m2, label="Double amplitude")
plt.plot(t*1000, m3, label="Double frequency")

plt.title("5.2-6: Message Signals")
plt.xlabel("Time (ms)")
plt.ylabel("m(t)")
plt.legend()
plt.grid()
plt.show()


# FM comparison
plt.figure()
plt.plot(t*1000, f_FM_1, label="Normal")
plt.plot(t*1000, f_FM_2, label="Double amplitude")
plt.plot(t*1000, f_FM_3, label="Double frequency")

plt.title("5.2-6: FM Frequency Deviation")
plt.xlabel("Time (ms)")
plt.ylabel("Frequency deviation (Hz)")
plt.legend()
plt.grid()
plt.show()


# PM comparison
plt.figure()
plt.plot(t*1000, f_PM_1, label="Normal")
plt.plot(t*1000, f_PM_2, label="Double amplitude")
plt.plot(t*1000, f_PM_3, label="Double frequency")

plt.title("5.2-6: PM Frequency Deviation")
plt.xlabel("Time (ms)")
plt.ylabel("Frequency deviation (Hz)")
plt.legend()
plt.grid()
plt.show()
```

```python
# ============================================================
# PROBLEM 5.2-7
# Gaussian message
# ============================================================

kf = 6000*np.pi
kp = 8000*np.pi

# Use a reasonable time scale
t = np.linspace(-3, 3, 20000)

# ------------------------------------------------------------
# Gaussian message
# ------------------------------------------------------------

m = np.exp(-t**2)


# ------------------------------------------------------------
# Derivative
# ------------------------------------------------------------

dm_dt = -2*t*np.exp(-t**2)


# ------------------------------------------------------------
# FM frequency deviation
# ------------------------------------------------------------

Delta_f_FM = kf*m/(2*np.pi)


# ------------------------------------------------------------
# PM frequency deviation
# ------------------------------------------------------------

Delta_f_PM = kp*dm_dt/(2*np.pi)


# ============================================================
# MESSAGE
# ============================================================

plt.figure()
plt.plot(t, m)

plt.title("5.2-7: Gaussian Message m(t) = exp(-t²)")
plt.xlabel("Time (s)")
plt.ylabel("m(t)")
plt.grid()
plt.show()


# ============================================================
# DERIVATIVE
# ============================================================

plt.figure()
plt.plot(t, dm_dt)

plt.title("5.2-7: Derivative dm(t)/dt")
plt.xlabel("Time (s)")
plt.ylabel("dm/dt")
plt.grid()
plt.show()


# ============================================================
# FM
# ============================================================

plt.figure()
plt.plot(t, Delta_f_FM)

plt.title("5.2-7: FM Frequency Deviation")
plt.xlabel("Time (s)")
plt.ylabel("Frequency deviation (Hz)")
plt.grid()
plt.show()


# ============================================================
# PM
# ============================================================

plt.figure()
plt.plot(t, Delta_f_PM)

plt.title("5.2-7: PM Frequency Deviation")
plt.xlabel("Time (s)")
plt.ylabel("Frequency deviation (Hz)")
plt.grid()
plt.show()
```

```python
# ============================================================
# PROBLEM 5.4-1
# PM -> FM demodulator -> Integrator
# ============================================================

t = np.linspace(0, 0.05, 10000)

fm = 100
wm = 2*np.pi*fm

# Original message
m = np.sin(wm*t)

# Output of FM demodulator when given a PM signal
demod_output = wm*np.cos(wm*t)

# Integrate
dt = t[1] - t[0]

recovered = np.cumsum(demod_output)*dt

# Remove initial constant caused by numerical integration
recovered = recovered - recovered[0]


plt.figure()
plt.plot(t, m, label="Original m(t)")
plt.plot(t, recovered/wm, "--", label="After demodulator + integrator")

plt.title("5.4-1: Recovering PM Message")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.show()
```

```python
# ============================================================
# 5.4-1(b)
# FM -> PM demodulator -> Differentiator
# ============================================================

t = np.linspace(0, 0.05, 10000)

fm = 100
wm = 2*np.pi*fm

# Original message
m = np.sin(wm*t)

# PM-demodulator output is proportional to integral of m
integral_m = -np.cos(wm*t)/wm

# Differentiate numerically
recovered = np.gradient(integral_m, t)


plt.figure()
plt.plot(t, m, label="Original m(t)")
plt.plot(t, recovered, "--", label="After PM demodulator + differentiator")

plt.title("5.4-1: Recovering FM Message")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.legend()
plt.grid()
plt.show()
```

```python
# ============================================================
# FM vs PM — MASTER VISUALIZATION
# ============================================================

fc = 10000
fm = 100

wc = 2*np.pi*fc
wm = 2*np.pi*fm

kf = 2*np.pi*1000
kp = 2

t = np.linspace(0, 0.05, 50000)

# Message
m = np.sin(wm*t)

# Derivative
dm_dt = wm*np.cos(wm*t)


# ------------------------------------------------------------
# FM instantaneous frequency
# ------------------------------------------------------------

f_FM = fc + (kf*m)/(2*np.pi)


# ------------------------------------------------------------
# PM instantaneous frequency
# ------------------------------------------------------------

f_PM = fc + (kp*dm_dt)/(2*np.pi)


# ------------------------------------------------------------
# Plot message
# ------------------------------------------------------------

plt.figure()
plt.plot(t, m)

plt.title("MASTER: Message m(t)")
plt.xlabel("Time (s)")
plt.ylabel("Amplitude")
plt.grid()
plt.show()


# ------------------------------------------------------------
# Plot FM frequency
# ------------------------------------------------------------

plt.figure()
plt.plot(t, f_FM)

plt.axhline(fc, linestyle="--")

plt.title("MASTER: FM Instantaneous Frequency")
plt.xlabel("Time (s)")
plt.ylabel("Frequency (Hz)")
plt.grid()
plt.show()


# ------------------------------------------------------------
# Plot PM frequency
# ------------------------------------------------------------

plt.figure()
plt.plot(t, f_PM)

plt.axhline(fc, linestyle="--")

plt.title("MASTER: PM Instantaneous Frequency")
plt.xlabel("Time (s)")
plt.ylabel("Frequency (Hz)")
plt.grid()
plt.show()
```
